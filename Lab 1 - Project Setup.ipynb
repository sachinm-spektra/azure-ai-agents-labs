In this lab you will set up your Microsoft Foundry resource and project, deploy essential language and embedding models, create necessary services, and setup necessary permissions and configurations.Finally, you’ll validate your setup by sending a test response request to your deployed model, ensuring your environment is ready for AI agent development in subsequent labs.

### Step 1: Test your lab environment

To test that your lab was setup successfully, run the below code that sends a message to the deployed model, asking it to tell a joke.

```python
# Import packages
import os
from azure.ai.projects import AIProjectClient
from azure.identity import DefaultAzureCredential
from dotenv import load_dotenv

load_dotenv() # Load environment variables from .env file
```

### Using token-based authentication using DefaultAzureCredential

This setup validation uses the Foundry OpenAI-compatible Responses endpoint with an API key from your `.env` file (`API_KEY` + `CHAT_MODEL_ENDPOINT`).

For the remaining labs (agent creation and runs), use token-based credentials (`AzureCliCredential` or `DefaultAzureCredential`) because the Agents SDK requires Microsoft Entra ID tokens.


```python
project = AIProjectClient(
    endpoint=os.getenv("AIPROJECT_ENDPOINT"),
    credential=DefaultAzureCredential()
)
```

### Chat with your model

This code calls the Foundry OpenAI-compatible `responses` API using your deployed chat model (`CHAT_MODEL`) and prints the response so you can verify your environment is configured correctly.

```python
# Chat with the model using the Responses API
# Make sure you run the above code cell to instantiate the project client first
with project.get_openai_client() as openai_client:
    response = openai_client.responses.create(
        model=os.getenv("CHAT_MODEL"),  # deployed model from your project
        instructions="You are an AI assistant that tells jokes for toddlers.",
        input="Hey, can you tell a joke about teddy bear?",
    )

print(response.output_text)
```
